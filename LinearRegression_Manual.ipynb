Goal is to do linear regression on the cars 24 dataset

```python
import pandas as pd
import numpy as np
data = pd.read_csv('train-cars24-car-price.csv')
data.head()
```

```python
data.columns
```

```python
# Using single feature to predict selling price of car
import seaborn as sns
import matplotlib.pyplot as plt
plt.figure(figsize=(10,6))
plt.title('Max power vs price')
sns.scatterplot(data=data, x='max_power', y='selling_price')
```

```python
def estimate_charges (maxpower, w, b):
  return w * maxpower + b
```

```python
w = 50
b = 100
```

```python
maxpower = data.max_power
estimated_charges = estimate_charges(maxpower, w, b)
estimated_charges
```

```python
plt.figure(figsize=(10,6))
plt.plot(maxpower, estimated_charges, 'y-o', label='Estimated_charges');
plt.xlabel('MaxPower')
plt.ylabel('Estimated Price')
plt.show()
```

```python
target = data.selling_price
plt.figure(figsize=(10,6))
plt.plot(maxpower, estimated_charges, 'y', alpha = 0.9);
plt.scatter(maxpower, target, s=8, alpha = 0.8)
plt.xlabel('MaxPower')
plt.ylabel('Price')
plt.legend(['Estimate', 'Actual'])
plt.show()
```

```python
def try_parameters(w,b):
  maxpower = data.max_power
  target = data.selling_price
  estimated_charges = estimate_charges(maxpower, w, b)
  plt.figure(figsize = (10,6))
  plt.plot(maxpower, estimated_charges, 'y', alpha = 0.9)
  plt.scatter(maxpower, target, s=8, alpha = 0.8)
  plt.xlabel('MaxPower')
  plt.ylabel('Price')
  plt.legend(['Estimate', 'Actual'])
  plt.show()
```

```python
try_parameters(50,100)
```

```python
try_parameters(0.5,1)
```

```python
try_parameters(0.08, 0.5)
```
